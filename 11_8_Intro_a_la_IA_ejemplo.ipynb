```python
# Ejemplo de sesgo en datos de entrenamniento

# Datos ficticios: (anios de experiencia, genero, contratado: 1 = SI, 0= NO
# Nota: "M" = masculino, "F" = femenino
datos = [
    (5, "M", 1),
    (4, "M", 1),
    (6, "M", 1),
    (5, "F", 0),
    (4, "F", 0),
    (6, "F", 0)
]

#Funcion simple para decidir si contratar (muy basica y sesgada)
def decidir_contratacion(anios_exp, genero):
    if genero == "M" and anios_exp >= 4:
        return "Contratado"
    else:
        return "No contratado"

# Probamos con dos cndidatos identicos pero distinto genero
print(decidir_contratacion(5, "M")) # Contratado
print(decidir_contratacion(5, "F")) # No contratado
```
