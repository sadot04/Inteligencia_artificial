```python
#Ejercicio 1: Perceptron simple (ADN logico)
import numpy as np
from sklearn.linear_model import Perceptron
#DATOS DE ENTRADA (X) y salida esperada (y)
#Entradas binarias (0 o 1)
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([0 , 0 , 0 , 1]) #Salida del AND logico
#Crear el perceptron
modelo = Perceptron(max_iter=1000, tol=1e-3)

#Entrenar
modelo.fit(X,y)

#Probar
print("Predicciones: ", modelo.predict(X))
print("Pesos: ", modelo.coef_)
print("Bias: ", modelo.intercept_)
```
